### Завдання 2.1
Виділіть кожну таблицю в окремий `DataFrame` — назвіть їх `A`, `B` і `C`.
Дайте стовпцям осмислені назви й перетворіть значення на числа
(`pd.to_numeric(..., errors='coerce')` стане в пригоді).

```python
# Таблиця A
A = RAW.iloc[14:, 1:15].copy()
A.columns = ['yr', 'mo', 'da', 'hhmm', 'mjd', 'secs', 'P>1', 'P>5', 'P>10', 'P>30', 'P>50', 'P>100', 'E>0.8', 'E>2.0']
A = A.apply(pd.to_numeric, errors='coerce').dropna(how='all')

# Таблиця B
B = RAW.iloc[14:, 16:20].copy()
B.columns = ['yr', 'mo', 'da', 'F10.7']
B = B.apply(pd.to_numeric, errors='coerce').dropna(how='all')

# Таблиця C
C = RAW.iloc[14:, 26:32].copy()
C.columns = ['yr', 'mo', 'da', 'hh', 'швидкість вітру', 'густина вітру']
C = C.apply(pd.to_numeric, errors='coerce').dropna(how='all')
```

### Завдання 2.2
У кожній таблиці дата й час записані **по-різному** й розкидані по кількох стовпцях.
Зберіть для кожної таблиці один стовпець `ts` типу `datetime`.

Після цього обов'язково виведіть для кожної таблиці найранішу й найпізнішу дату.

*Очікуваний результат: усі три таблиці мають лежати в межах серпня–вересня 2017 року.
Якщо у вас вийшов інший рік — придивіться, як записано рік у тій таблиці, і виправте.*

```python
# Таблиця A
A_time = A['yr'].astype(int).astype(str) + '-' + A['mo'].astype(int).astype(str) + '-' + A['da'].astype(int).astype(str) + ' ' + (A['hhmm'] // 100).astype(int).astype(str).str.zfill(2) + ':' + (A['hhmm'] % 100).astype(int).astype(str).str.zfill(2)
A['ts'] = pd.to_datetime(A_time, errors='coerce')

# Таблиця B (готові колонки року, місяця, дня)
B['ts'] = pd.to_datetime(B[['yr', 'mo', 'da']].rename(columns={'yr':'year', 'mo':'month', 'da':'day'}), errors='coerce')

# Таблиця C (рік 17 - 2017)
C_time = (C['yr'] + 2000).astype(int).astype(str) + '-' + C['mo'].astype(int).astype(str) + '-' + C['da'].astype(int).astype(str) + ' ' + C['hh'].astype(int).astype(str).str.zfill(2) + ':00'
C['ts'] = pd.to_datetime(C_time, errors='coerce')

print("A:", A['ts'].min(), "—", A['ts'].max())
print("B:", B['ts'].min(), "—", B['ts'].max())
print("C:", C['ts'].min(), "—", C['ts'].max())
```

**Зверніть увагу:** таблиці закінчуються не в один і той самий момент. Тому результат

```python
C_clean = C[['ts', 'швидкість вітру', 'густина вітру']].dropna(subset=['ts'])
B_clean = B[['ts', 'F10.7']].dropna(subset=['ts'])
B_clean = B_clean.rename(columns={'ts': 'date'})

M = pd.merge(A_hourly, C_clean, on='ts', how='inner')
M['date'] = M['ts'].dt.normalize()
M = pd.merge(M, B_clean, on='date', how='inner').drop(columns=['date'])

print(M.shape)
display(M[M.isna().any(axis=1)])
```

Оскільки нам потрібно аналізувати зв'язки між цими параметрами, рядки з порожніми значеннями нам не підходять.

### Завдання 4.4 — поділ на групи
Щоб далі було що з чим порівнювати, поділіть години на три групи за значенням
**максимуму потоку P>10 за годину**:

| Група | Умова |
|---|---|
| `тиха` | менше 10 |
| `активна` | від 10 до 1000 |
| `дуже активна` | 1000 і більше |

Додайте до `M` стовпець `група` і виведіть, скільки годин потрапило в кожну.

> Це не вигадані числа: саме такі пороги використовує метеослужба космічної погоди
> NOAA. Чому вони саме такі — питання не сьогоднішньої роботи.

```python
def p10_group(val):
    if pd.isna(val): return np.nan
    if val < 10: return 'тиха'
    if val < 1000: return 'активна'
    return 'дуже активна'

M['група'] = M['P>10_max'].apply(p10_group)
print(M['група'].value_counts())
```
